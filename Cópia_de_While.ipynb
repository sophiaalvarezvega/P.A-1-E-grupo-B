```python
numero = int(input("Digite um número: "))
contador = 1

while contador <= numero:
    print(contador)
    contador += 1
```

```python
numero = int(input("Digite um número: "))

while numero >= 0:
    print(numero)
    numero -= 1
```

```python
numero = int(input("Digite um número: "))
contador = 1
soma = 0

while contador <= numero:
    soma += contador
    contador += 1

print(f"Soma: {soma}")
```

```python
numero = int(input("Digite um número: "))
contador = 1

while contador <= 10:
    print(f"{numero} x {contador} = {numero * contador}")
    contador += 1
```

```python
senha = input("Digite a senha: ")

while senha != "1234":
    print("Senha incorreta")
    senha = input("Digite a senha novamente: ")

print("Acesso permitido")
```

```python
soma = 0
numero = int(input("Digite um número: "))

while numero != 0:
    soma += numero
    numero = int(input("Digite um número: "))

print(f"Soma: {soma}")
```

```python
numero = int(input("Digite um número: "))
maior = numero

while numero != 0:
    if numero > maior:
        maior = numero
    numero = int(input("Digite um número: "))

print(f"Maior número: {maior}")
```

```python
contador = 1
soma = 0

while contador <= 5:
    nota = float(input("Digite a nota: "))
    soma = soma + nota
    contador = contador + 1

media = soma / 5
print("Média:", media)
```

```python
saldo = 1000
opcao = 0

while opcao != 4:
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")
    opcao = int(input("Opção: "))

    if opcao == 1:
        valor = float(input("Valor do depósito: "))
        saldo = saldo + valor
    if opcao == 2:
        valor = float(input("Valor do saque: "))
        saldo = saldo - valor
    if opcao == 3:
        print("Saldo:", saldo)

print("Fim do programa")
```
