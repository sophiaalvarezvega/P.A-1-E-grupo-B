```
#Cofre digital

#um cofre possui um codigo composto por 5 números inteiros.

#desenvolva um programa que:

#leia os números utilizando um laço de repetição.

# ao final informar:

# quantos numeros sao pares
# quantos sao impares
# o maior numero
# a soma de todos os numeros
# a media deles
# o menor numero
#quantos numeros sao maiores que a média

pares = 0
impares = 0
soma = 0

for i in range(5):
    n = int(input("Digite um número: "))

    if i == 0:
        maior = n
        menor = n

    if n > maior:
        maior = n

    if n < menor:
        menor = n

    if n % 2 == 0:
        pares = pares + 1
    else:
        impares = impares + 1

    soma = soma + n

    if i == 0:
        n1 = n
    elif i == 1:
        n2 = n
    elif i == 2:
        n3 = n
    elif i == 3:
        n4 = n
    else:
        n5 = n

media = soma / 5

acima_media = 0

for n in (n1, n2, n3, n4, n5):
    if n > media:
        acima_media = acima_media + 1

print("Pares:", pares)
print("Ímpares:", impares)
print("Maior número:", maior)
print("Menor número:", menor)
print("Soma:", soma)
print("Média:", media)
print("Números maiores que a média:", acima_media)
```

```
# numero espelho

#solicite um numero inteiro positivo

#inverta o numero
#informe se ele é um palindromo

numero = int(input("Digite um número: "))

copia = numero
espelho = 0

while copia > 0:
    espelho = espelho * 10 + copia % 10
    copia = copia // 10

print("numero espelho:", espelho)

if numero == espelho:
    print("É um palíndromo.")
else:
    print("Não é um palíndromo.")
```
