```python
cor = 'azul'
match cor:
  case 'azul':
    print('Cor azul')
  case _:
        print("Outra cor")
```

```python
numero = 2

match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
    case _:
        print("Número inválido")
```

```python
bebida = 2

match bebida:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")
```

```python
dia = "sábado"

match dia:
    case "sábado":
        print("Final de semana")
    case "domingo":
        print("Final de semana")
    case _:
        print("Dia de semana")
```

```python
nota = "A"

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```
